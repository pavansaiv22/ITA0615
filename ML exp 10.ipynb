```python
from sklearn.mixture import GaussianMixture
import numpy as np

# Sample dataset
X = np.array([[1], [2], [3], [10], [11], [12]])

# EM Algorithm using Gaussian Mixture Model
gmm = GaussianMixture(n_components=2, random_state=0)

# Train model
gmm.fit(X)

# Predict clusters
labels = gmm.predict(X)

print("Cluster Labels:")
print(labels)
```
